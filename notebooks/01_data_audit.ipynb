# 01. Data Audit & Exploratory Data Analysis
Auditing the raw parallel datasets from Ghana-NLP for Twi and Ewe.

```python
import sys
sys.path.append("..")
from src.preprocessing import download_raw

# Download and load raw datasets
twi_df = download_raw("Ghana-NLP/ENGLISH_TWI_PARALLEL_TEXT", "twi", text_column="label")
ewe_df = download_raw("Ghana-NLP/EWE_ENGLISH_PARALLEL_TEXT", "ewe", text_column="label")
```

```python
for name, df in [("Twi", twi_df), ("Ewe", ewe_df)]:
    print(f"\n--- {name} ---")
    print("Columns:", list(df.columns))
    print("Rows:", len(df))
    print("Missing values per column:\n", df.isna().sum())
    print("Duplicate rows:", df.duplicated().sum())
    print("Sample rows:")
    print(df.sample(5, random_state=42))
```

### Data Audit Observations & Findings:
1. **Column Mapping & Artifacts**: In both datasets, the English text is stored in `text` and the Ghanaian target language is stored in `label`. The Twi dataset contains an artifact column `Unnamed: 3` with almost all nulls (6,087 / 6,090).
2. **Missing & Incomplete Rows**: There are missing values in the target translation `label` column (3 missing in Twi, 27 missing in Ewe) which must be dropped during preprocessing.
3. **Orthography & Special Characters**: Twi contains unique open vowels (`ɛ`, `ɔ`), while Ewe features rich phonetic and tone orthography (`ɖ`, `ƒ`, `ŋ`, `ɔ`, `ɛ`, `ʋ`, `ɣ`, and diacritics like `lɔlɔ̃tɔa`). Average sentence lengths are typical for parallel translation tasks (~10-25 words per sentence).
