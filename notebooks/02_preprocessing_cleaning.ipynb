# 02. Preprocessing & Data Splitting
Cleans raw Twi and Ewe corpora and splits them into train (80%), validation (10%), and test (10%) sets.

```python
import sys
sys.path.append("..")
import pandas as pd
from src.preprocessing import clean_dataframe, split_and_save

twi_df = pd.read_csv("../data/raw/twi/raw_sentences.csv")
ewe_df = pd.read_csv("../data/raw/ewe/raw_sentences.csv")

twi_clean = clean_dataframe(twi_df, text_column="twi")
ewe_clean = clean_dataframe(ewe_df, text_column="ewe")

print("Twi: kept", len(twi_clean), "of", len(twi_df))
print("Ewe: kept", len(ewe_clean), "of", len(ewe_df))

split_and_save(twi_clean, text_column="twi", language="twi")
split_and_save(ewe_clean, text_column="ewe", language="ewe")
```
